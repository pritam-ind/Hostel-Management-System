Hostel Management System

```python
out_alloted_ac_rooms = set()
out_alloted_non_ac_rooms = set()

inside_alloted_ac_rooms = set()
inside_alloted_non_ac_rooms = set()

roomCosting = {"AC Single" : 160000,
              "AC Double" : 140000,
              "NON-AC Single" : 120000,
              "NON-AC Double" : 110000,
              "NON-AC Triple" : 900000}
```

```python
def print_ac_type_cost(room_dict):   
    found = False
    for room_type, cost in room_dict.items():
        if "AC" in room_type and not "NON-AC" in room_type:
            print(f"{room_type} : ₹{cost}")
            found = True
        # elif not found:
        #     print("No AC room found")
    print()

def print_non_ac_type_cost(room_dict):
    found = True
    for room_type, cost in room_dict.items():
        if "NON-AC" in room_type:
            print(f"{room_type} : ₹{cost}")
            found = False
        # elif found:
        #     print("No NON-AC room found")
    print()


def pay_ac_single_double(choose):
    if choose == "single":
        print("\nEnter your upi here: ", end=" ")
        upi = input()
        print()
        print("A payment link of ₹160000 has send to your number. \nPlease pay that amount immediatly befor expire of link\n")
        print("\nThanks for the payment. Your room has been alloted")
    elif choose == "double":
        print("\nEnter your upi here: ", end=" ")
        upi = input()
        print()
        print("A payment link of ₹140000 has send to your number. \nPlease pay that amount immediatly befor expire of link\n")
        print("\nThanks for the payment. Your room has been alloted")
    else:
        print("\nPlease enter a valid input!")


def pay_non_ac_single_double_triple(choose):
    if choose == "single":
        print("\nEnter your upi here: ", end=" ")
        upi = input("\n")
        print()
        print("A payment link of ₹120000 has send to your number. \nPlease pay that amount immediatly befor expire of link\n")
        print("\nThanks for the payment. Your room has been alloted")
    elif choose == "double":
        print("\nEnter your upi here: ", end=" ")
        upi = input("\n")
        print()
        print("A payment link of ₹110000 has send to your number. \nPlease pay that amount immediatly befor expire of link\n")
        print("\nThanks for the payment. Your room has been alloted")
    elif choose == "triple":
        print("\nEnter your upi here: ", end=" ")
        upi = input("\n")
        print()
        print("A payment link of ₹90000 has send to your number. \nPlease pay that amount immediatly befor expire of link\n")
        print("\nThanks for the payment. Your room has been alloted\n")
    else:
        print("\nPlease enter a valid input!")
```

```python
# import random as r
#booking the hostel if room is not alloted
flag = True
while flag:
    chooseCampus = input("which hostel do you want to choose: (inside/outside)").lower().strip()
    #if student wish to choose the inside Hostel
    if chooseCampus == "inside":
        isAC = input("Select the type AC/NON-AC: ").lower().strip()
        # AC
        if isAC=="ac":
            #printing the type and cost of ac rooms usign the function
            print("AC Room and Respective Cost\n")
            print(print_ac_type_cost(roomCosting))

            #payment gateway process
            choose = input("\nChoose Single or Double AC rooms ").lower().strip()
            if choose == "single" or choose == "double":
                #calling the payment fun for AC rooms
                pay_ac_single_double(choose)
              
            else:
                print("\nplease select a valid input")
            

        # NON-AC  
        elif isAC == "non-ac":
            #printing the type and cost of non-ac rooms using the function
            print(print_non_ac_type_cost(roomCosting)) 

            #payment gateway process
            choose = input("\nChoose Single or Double or Triple NON-AC rooms ").lower().strip()
            if choose == "single" or choose == "double" or choose =="triple":
                #calling the payment fun for NON-AC rooms
                pay_non_ac_single_double_double(choose)
              
            else:
                print("\nplease select a valid input")

            

        
    #if student wish to choose the ouside Hostel
    elif chooseCampus == "outside":
        isAC = input("Select the type AC/NON-AC: ").lower().strip()
        if isAC=="ac":
            #printing the type and cost of ac rooms usign the function
            print("NON-AC Room and Respective Cost\n")
            print(print_ac_type_cost(roomCosting))

            #payment gateway process
            choose = input("\nChoose Single or Double AC rooms ").lower().strip()
            if choose == "single" or choose == "double":
                #calling the payment fun for AC rooms
                pay_ac_single_double(choose)
              
            else:
                print("\nplease select a valid input")
            
                
        elif isAC == "non-ac":
            #printing the type and cost of non-ac rooms using the function
            print(print_non_ac_type_cost(roomCosting))

            #payment gateway process
            choose = input("\nChoose Single or Double or Triple NON-AC rooms ").lower().strip()
            if choose == "single" or choose == "double" or choose =="triple":
                #calling the payment fun for NON-AC rooms
                pay_non_ac_single_double_triple(choose)
              
            else:
                print("\nplease select a valid input")

        else:
            print("Please enter a valid input")


        
```
