HOSTEL MANAGEMENT SYSTEM

```python
# --- DATA STRUCTURES INITIALIZATION ---
occupied_rooms = {101, 102}  # Set: Unique rooms already taken
student_db = {               # Dictionary: Student Details
    "VGU001": {"name": "Amit", "room": 101},
    "VGU002": {"name": "Rahul", "room": 102}
}
complaint_log = ["Wifi issue in 101"] # List: Mutable log of issues

# --- MAIN PROGRAM LOOP ---
while True:
    print("\n--- VGU HOSTEL MANAGER ---")
    print("1. Allot New Room")
    print("2. View Student Details")
    print("3. Log a Complaint (List)")
    print("4. Save & Exit")
    
    choice = input("Enter choice (1-4): ")

    # --- OPTION 1: ALLOTMENT (Set & Tuple) ---
    if choice == '1':
        room_num = int(input("Enter Room Number to Allot: "))
        
        # Logic: Check if room is in the Set
        if room_num in occupied_rooms:
            print(f"Sorry, Room {room_num} is already occupied!")
        else:
            # Taking Input
            roll_no = input("Enter Roll Number: ")
            name = input("Enter Student Name: ")
            room_type = input("AC or Non-AC? ")

            # 1. Add to Set (To prevents duplicates later)
            occupied_rooms.add(room_num)

            # 2. Add to Dictionary (To store details)
            student_db[roll_no] = {"name": name, "room": room_num}

            # 3. Create a Tuple (Immutable Receipt)
            # Once created, this specific booking record cannot be edited
            booking_receipt = (roll_no, name, room_num, room_type)
            
            print("Room Allotted Successfully!")
            print(f"Booking Receipt: {booking_receipt}")

    # --- OPTION 2: VIEW DETAILS (Dictionary) ---
    elif choice == '2':
        roll_check = input("Enter Roll Number to search: ")
        # Logic: Check if key exists in Dictionary
        if roll_check in student_db:
            print("Found:", student_db[roll_check])
        else:
            print("Student not found.")

    # --- OPTION 3: COMPLAINTS (List) ---
    elif choice == '3':
        # Lists are mutable, so we can append new issues
        issue = input("Enter the complaint (e.g., Fan not working in 105): ")
        complaint_log.append(issue) 
        print("Complaint added. Current Log:")
        print(complaint_log)

    # --- OPTION 4: FILE HANDLING ---
    elif choice == '4':
        # using 'with' statement to automatically close file 
        # mode 'a' (append) or 'w' (write) can be used. 
        with open("hostel_records.txt", "w") as file:
            file.write("--- CURRENT OCCUPANCY ---\n")
            file.write(str(student_db) + "\n") # Converting dict to string to save it
            
            file.write("\n--- COMPLAINT LOG ---\n")
            for item in complaint_log:
                file.write(item + "\n")
                
        print("Data saved to 'hostel_records.txt'. Exiting...")
        break

    else:
        print("Invalid choice. Try again.")
```

REAL-TIME ROOM ALLOTMENT SYSTEM
Tuple: Represents the Physical Hostel (These rooms exist and don't change).

Set: Represents Reality (Which rooms are locked/occupied).

List: This is the Key Solver. We will create a "Walk-in List" dynamically.

Logic: Total Rooms (Tuple) - Occupied Rooms (Set) = Available List.

File Handling: Saves the Set immediately so the next admin sees the update instantly.

```python
import os

# TUPLE: (Room Number, Floor, Type)
HOSTEL_INVENTORY = (
    (120, 1, "AC"),
    (130, 1, "Non-AC"),
    (210, 2, "AC"),
    (229, 2, "Non-AC"),
    (301, 3, "AC"),
    (302, 3, "Non-AC")
)

#set...no duplication allowed and faster than a list
occupied_rooms_set = set()

# DICTIONARY: Student Details 
# Key=Room Number, Value=Student Name
student_data = {}

# FILE HANDLING: it loads the previous data, provides the current status
file_name = "vgu_hostel_status.txt"

if os.path.exists(file_name):
    print("Loading existing data...")
    with open(file_name, "r") as file:
        for line in file:
            # We assume file format is: Room_No,Student_Name
            # strip() removes spaces/newlines, split(',') separates data
            data = line.strip().split(',')
            
            # Safety check to ensure line has data
            if len(data) >= 2:
                room_num = int(data[0])
                student_name = data[1]
                
                # Load into memory
                occupied_rooms_set.add(room_num)
                student_data[room_num] = student_name
    print("Data loaded successfully.")

# main program loops
while True:
    print("\n--- VGU REAL-TIME ALLOTMENT SYSTEM ---")
    print("1. Show Available Rooms (The Fix)")
    print("2. Book a Room")
    print("3. View Occupant Details")
    print("4. Exit")
    
    choice = input("Enter choice (1-4): ")

    # option 1
    if choice == '1':
        available_rooms_list = []
        
        print("\nScanning availability...")
        for room in HOSTEL_INVENTORY:
            r_num = room[0]
            r_type = room[2]
            
            # LOGIC: Only add to list if NOT in the Occupied Set
            if r_num not in occupied_rooms_set:
                available_rooms_list.append(room)
        
        # Display list
        if not available_rooms_list:
            print("ALERT: No rooms available!")
        else:
            print(f"\nFound {len(available_rooms_list)} available rooms:")
            for room in available_rooms_list:
                print(f" -> Room {room[0]} ({room[2]}) - Floor {room[1]}")

    # option2: Booking process
    elif choice == '2':
        try:
            target_room = int(input("Enter Room Number to Allot: "))
            
            if target_room in occupied_rooms_set:
                print(f"ERROR: Room {target_room} is ALREADY OCCUPIED by {student_data.get(target_room)}.")
            
            #  To check (Does room exist?)
            else:
                # Check if the number entered is actually a valid room in our Tuple
                valid_room = False
                for r in HOSTEL_INVENTORY:
                    if r[0] == target_room:
                        valid_room = True
                        break
                
                if valid_room:
                    s_name = input("Enter Student Name: ")
                    
                    # Update Memory
                    occupied_rooms_set.add(target_room)
                    student_data[target_room] = s_name
                    
                    # Update File IMMEDIATELY
                    # using 'a' (append) mode so we don't delete old records
                    with open(file_name, "a") as file:
                        file.write(f"{target_room},{s_name}\n")
                        
                    print(f"Success! Room {target_room} has been locked for {s_name}.")
                else:
                    print("Error: This room number does not exist in the hostel map.")

        except ValueError:
            print("Invalid Input! Please enter a number for the room.")

    # option 3: viewing the details
    elif choice == '3':
        try:
            search_room = int(input("Enter Room Number to check: "))
            if search_room in student_data:
                print(f"Room {search_room} is occupied by: {student_data[search_room]}")
            else:
                print("This room is currently empty.")
        except ValueError:
            print("Please enter a valid number.")

    # Option 4: Exit
    elif choice == '4':
        print("Exiting System...")
        break

    else:
        print("Invalid choice. Please try again.")
```
