```python
import pyttsx3
import time # Import the time module

speaker = pyttsx3.init()
names = ["pritam", "sachin", "souvik", "aditya", "ayush", "beta sumit"]

for name in names:
    phrase = f"hello {name} sir, how's your internship going on!"
    print(phrase) # Add a print statement to see the script progressing
    speaker.say(phrase)
    speaker.runAndWait() # Wait for the current phrase to finish
    time.sleep(0.5) # Add a 500ms (half-second) pause between names
```

```python
# speaking multiple times
import pyttsx3
import time

names = ["pritam", "sachin", "souvik", "aditya", "ayush", "beta sumit"]

for name in names:
    # 1. INITIALIZE the engine inside the loop (start fresh)
    speaker = pyttsx3.init()
    
    phrase = f"hello {name} sir, how's your internship going on!"
    print(phrase) # Keep the print for visual confirmation
    
    # 2. SAY and WAIT
    speaker.say(phrase)
    speaker.runAndWait()
    
    # 3. STOP and DELETE the engine (clean up resources)
    speaker.stop()
    del speaker

    # Optional: A small pause can sometimes help on slow systems
    time.sleep(1.0)
    
print("Loop finished.")
```
